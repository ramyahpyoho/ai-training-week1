```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../task 1/clean_data.csv")

print("Dataset loaded successfully!")
df.head()
```

```python
print("Dataset Shape;",df.shape)
print(df)
```

```python
department_salary = df.groupby("department")["salary"].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    department_salary.index,
    department_salary.values,
    color="skyblue",
    edgecolor="black"
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (₹)")

plt.tight_layout()

# Save BEFORE show
plt.savefig(
    "reports/figures/bar_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Bar chart saved successfully!")
```

1.Bar chart
The Finance department has the highest average salary of ₹90,000, while Sales has the lowest average salary of ₹41,000.

```python
# Save bar chart
plt.savefig("reports/figures/bar_chart.png", dpi=300, bbox_inches="tight")

print("Bar chart saved successfully!")
```

```python
# Create histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df["salary"],
    bins=5,
    color="orange",
    edgecolor="black"
)

plt.title("Employee Salary Distribution")
plt.xlabel("Salary (₹)")
plt.ylabel("Number of Employees")

plt.tight_layout()

plt.savefig(
    "reports/figures/histogram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Histogram saved successfully!")
```

2. Histogram
Employee salaries are distributed across a range of ₹30,000 to ₹90,000, with most salaries falling between ₹30,000 and ₹75,000.

```python

plt.savefig("reports/figures/histogram.png",
            dpi=300, bbox_inches="tight")

print("Histogram saved successfully!")
```

```python
# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df["id"],
    df["salary"],
    color="green",
    edgecolor="black",
    s=80
)

plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary (₹)")

plt.tight_layout()

plt.savefig(
    "reports/figures/scatter_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Scatter plot saved successfully!")
```

3. Scatter plot
The scatter plot shows a weak positive relationship between employee ID and salary, but employee ID is not a meaningful business measure.

```python
plt.savefig("reports/figures/scatter_plot.png",
            dpi=300, bbox_inches="tight")

print("Scatter plot saved successfully!")
```

```python
sorted_salary = df["salary"].sort_values()

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(sorted_salary) + 1),
    sorted_salary,
    marker="o",
    color="purple",
    linewidth=2
)

plt.title("Employee Salaries in Ascending Order")
plt.xlabel("Employee Rank")
plt.ylabel("Salary (₹)")

plt.tight_layout()

plt.savefig(
    "reports/figures/line_chart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Line chart saved successfully!")
```

4. Line chart
The sorted salary chart shows that employee salaries increase from ₹30,000 to ₹90,000.

```python
plt.savefig("reports/figures/line_chart.png",
            dpi=300, bbox_inches="tight")

print("Line chart saved successfully!")
```

```python

plt.figure(figsize=(8, 5))

plt.boxplot(
    df["salary"],
    patch_artist=True,
    boxprops=dict(facecolor="lightblue")
)

plt.title("Employee Salary Box Plot")
plt.xlabel("Employees")
plt.ylabel("Salary (₹)")

plt.tight_layout()

plt.savefig(
    "reports/figures/box_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Box plot saved successfully!")
```

5. Box plot
The box plot shows the spread of employee salaries, with no outliers identified using the IQR method.

```python

plt.savefig("reports/figures/box_plot.png",
            dpi=300, bbox_inches="tight")

print("Box plot saved successfully!")
```
