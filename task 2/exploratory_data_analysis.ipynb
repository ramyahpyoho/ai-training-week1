```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset from Task 1
df = pd.read_csv("../task 1/clean_data.csv")

print("Cleaned dataset loaded successfully!")
df.head()
```

```python
# Calculate salary statistics
print("Salary Statistics:")

print("Mean:", df["salary"].mean())
print("Median:", df["salary"].median())
print("Minimum:", df["salary"].min())
print("Maximum:", df["salary"].max())
print("Standard Deviation:", df["salary"].std())
```

```python
# Department-wise salary analysis
department_analysis = df.groupby("department")["salary"].agg(
    ["mean", "min", "max", "count"]
)

print("Department-wise Salary Analysis:")
department_analysis
```

```python
# City-wise salary analysis
city_analysis = df.groupby("city")["salary"].agg(
    ["mean", "min", "max", "count"]
)

print("City-wise Salary Analysis:")
city_analysis
```

```python
# Calculate correlation between numeric columns
correlation = df[["id", "salary"]].corr()

print("Correlation Analysis:")
correlation
```

```python
# Create department budget table
budget = pd.DataFrame({
    "department": ["Sales", "HR", "IT", "Finance"],
    "budget": [100000, 80000, 150000, 120000]
})

print("Department Budget:")
budget
```

```python
# Merge employee data with department budget
merged_df = pd.merge(df, budget, on="department", how="left")

print("Merged Dataset:")
merged_df
```

```python
# Standardize department names in budget table
budget["department"] = budget["department"].str.title()

# Merge again
merged_df = pd.merge(df, budget, on="department", how="left")

print("Corrected Merged Dataset:")
merged_df
```

```python
# Create pivot table
pivot_table = pd.pivot_table(
    df,
    values="salary",
    index="department",
    columns="city",
    aggfunc="mean"
)

print("Department and City Salary Pivot Table:")
pivot_table
```
