```python
df = pd.read_csv("../employees.csv", sep="\t")

print("Dataset loaded correctly!")
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Check data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Remove unwanted spaces and standardize text
df["name"] = df["name"].str.strip().str.title()
df["department"] = df["department"].str.strip().str.title()
df["city"] = df["city"].str.strip().str.title()

# Display cleaned dataset
df
```

```python
# Calculate Q1 and Q3
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Find lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# Identify outliers
outliers = df[
    (df["salary"] < lower_limit) |
    (df["salary"] > upper_limit)
]

print("\nOutliers:")
print(outliers)
```

```python
# Create before and after cleaning summary
summary = pd.DataFrame({
    "Stage": ["Before Cleaning", "After Cleaning"],
    "Rows": [8, len(df)],
    "Missing Values": [0, df.isnull().sum().sum()],
    "Duplicates": [0, df.duplicated().sum()]
})

summary
```

```python
# Save cleaned dataset
df.to_csv("clean_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
